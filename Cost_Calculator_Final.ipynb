```python
import math

# =========================================================
# Pakistan Parametric Construction Cost Engine (v3)
# PRODUCTION VERSION - Reads ALL rates from Config (Cell 0)
# =========================================================

# IMPORTANT: Cell 0 (Config Manager) must be run before this cell!

ENGINE_METADATA = {
    "engine": "Pakistan Parametric Cost Engine",
    "version": "3.0",
    "currency": "PKR",
    "area_unit": "sqft",
    "target_market": "Well-developed housing societies (DHA, Bahria, Gulberg, etc.)"
}


# -----------------------
# ALL CONFIGURATION FROM CELL 0 (No hardcoded values)
# -----------------------

# Get configuration from Cell 0
config = Config.get_config()

# Plot rules
PLOT_RULES = config["plot_rules"]

# City factors
CITY_FACTOR = config["city_factors"]

# Soil factors
SOIL_FACTOR = config["soil_factors"]

# Quality multipliers
QUALITY_MULTIPLIER = config["quality_multipliers"]

# Labour & Contractor Rates
LABOUR_RATE_GREY = config["labour_rates"]["grey_structure"]
LABOUR_RATE_FINISH = config["labour_rates"]["finishing"]
GREY_CONTRACTOR_FEE = config["contractor_fees"]["grey_structure"]
FINISH_CONTRACTOR_FEE = config["contractor_fees"]["finishing"]

# Base consumption norms
BASE_NORMS = {
    # Grey structure
    "steel_kg": config["base_consumption"]["grey_structure"]["steel_kg"],
    "cement_bags": config["base_consumption"]["grey_structure"]["cement_bags"],
    "sand_cft": config["base_consumption"]["grey_structure"]["sand_cft"],
    "crush_cft": config["base_consumption"]["grey_structure"]["crush_cft"],
    "bricks": config["base_consumption"]["grey_structure"]["bricks"],
    # Finishes
    "wiring_points": config["base_consumption"]["finishing"]["wiring_points"],
    "plumbing_points": config["base_consumption"]["finishing"]["plumbing_points"],
    "tiles_sqft": config["base_consumption"]["finishing"]["tiles_sqft"],
    "paint_sqft": config["base_consumption"]["finishing"]["paint_sqft"],
    "ceiling_sqft": config["base_consumption"]["finishing"]["ceiling_sqft"],
    "wood_sqft": config["base_consumption"]["finishing"]["wood_sqft"]
}

# Room multipliers
ROOM_MULTIPLIERS = config["room_multipliers"]

# Add-on multipliers
ADDON_MULTIPLIERS = config["addon_multipliers"]

# Material prices
MATERIAL_PRICES = config["material_prices"]

# Solar Systems
SOLAR_SYSTEMS = config["solar_systems"]

# Insulation rates
INSULATION_RATE = config["insulation_rates"]


# =========================================================
# Helper Functions
# =========================================================

def calculate_room_multipliers(bedrooms, washrooms, living, floors):
    """Calculate material additions based on room configuration"""

    extra_bedrooms = max(0, bedrooms - 2)
    extra_living = max(0, living - 1)
    extra_floors = max(0, floors - 1)

    multipliers = {
        "steel_kg": 0, "cement_bags": 0, "sand_cft": 0, "crush_cft": 0,
        "bricks": 0, "wiring_points": 0, "plumbing_points": 0,
        "tiles_sqft": 0, "paint_sqft": 0, "wood_sqft": 0
    }

    # Add per-extra-bedroom multipliers
    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_extra_bedroom"]:
            multipliers[key] += extra_bedrooms * ROOM_MULTIPLIERS["per_extra_bedroom"][key]

    # Add per-washroom multipliers
    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_washroom"]:
            multipliers[key] += washrooms * ROOM_MULTIPLIERS["per_washroom"][key]

    # Add per-extra-living multipliers
    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_extra_living"]:
            multipliers[key] += extra_living * ROOM_MULTIPLIERS["per_extra_living"][key]

    # Add per-extra-floor multipliers
    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_extra_floor"]:
            multipliers[key] += extra_floors * ROOM_MULTIPLIERS["per_extra_floor"][key]

    return multipliers, extra_floors, extra_bedrooms, extra_living


def calculate_addon_impacts(has_porch, has_balcony, base_area):
    """Calculate material additions from porch and balcony"""

    porch_area = 0
    balcony_area = 0
    addon_multipliers = {
        "steel_kg": 0, "cement_bags": 0, "sand_cft": 0, "crush_cft": 0,
        "tiles_sqft": 0, "paint_sqft": 0, "metal_glass_cost": 0,
        "waterproofing_multiplier": 0
    }

    if has_porch:
        porch_area = base_area * ADDON_MULTIPLIERS["porch"]["area_multiplier"]
        addon_multipliers["steel_kg"] += porch_area * ADDON_MULTIPLIERS["porch"]["steel_kg"]
        addon_multipliers["cement_bags"] += porch_area * ADDON_MULTIPLIERS["porch"]["cement_bags"]
        addon_multipliers["sand_cft"] += porch_area * ADDON_MULTIPLIERS["porch"]["sand_cft"]
        addon_multipliers["crush_cft"] += porch_area * ADDON_MULTIPLIERS["porch"]["crush_cft"]
        addon_multipliers["tiles_sqft"] += porch_area * ADDON_MULTIPLIERS["porch"]["tiles_sqft"]
        addon_multipliers["paint_sqft"] += porch_area * ADDON_MULTIPLIERS["porch"]["paint_sqft"]
        addon_multipliers["metal_glass_cost"] += porch_area * ADDON_MULTIPLIERS["porch"]["metal_glass"]
        addon_multipliers["waterproofing_multiplier"] += ADDON_MULTIPLIERS["porch"]["waterproofing"]

    if has_balcony:
        balcony_area = base_area * ADDON_MULTIPLIERS["balcony"]["area_multiplier"]
        addon_multipliers["steel_kg"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["steel_kg"]
        addon_multipliers["cement_bags"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["cement_bags"]
        addon_multipliers["sand_cft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["sand_cft"]
        addon_multipliers["crush_cft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["crush_cft"]
        addon_multipliers["tiles_sqft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["tiles_sqft"]
        addon_multipliers["paint_sqft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["paint_sqft"]
        addon_multipliers["metal_glass_cost"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["metal_glass"]
        addon_multipliers["waterproofing_multiplier"] += ADDON_MULTIPLIERS["balcony"]["waterproofing"]

    total_addon_area = porch_area + balcony_area

    return addon_multipliers, total_addon_area


# =========================================================
# Core estimator
# =========================================================

def estimate_house(
    plot,
    floors,
    bedrooms,
    washrooms,
    living,
    city,
    soil_type,
    quality_level,
    brands,
    porch=False,
    balcony=False,
    solar=False,
    insulation=False,
    insulation_type=None,
    ceiling_type=None
):

    if plot not in PLOT_RULES:
        raise ValueError("Invalid plot size")

    rule = PLOT_RULES[plot]

    # Validate totals against plot limits
    if bedrooms > rule["max_bed_total"]:
        raise ValueError(f"Bedrooms ({bedrooms}) exceed plot limit ({rule['max_bed_total']})")

    if washrooms > rule["max_wash"]:
        raise ValueError(f"Washrooms ({washrooms}) exceed plot limit ({rule['max_wash']})")

    if living > rule["max_living_total"]:
        raise ValueError(f"Living rooms ({living}) exceed plot limit ({rule['max_living_total']})")

    # Calculate base area
    plot_sqft = rule["plot_sqft"]
    coverage = rule["coverage"]
    covered_per_floor = plot_sqft * coverage
    base_area = covered_per_floor * floors

    # Apply add-ons area impact
    addon_impacts, addon_area = calculate_addon_impacts(porch, balcony, base_area)
    total_covered = base_area + addon_area

    # Get room-based multipliers
    room_multipliers, extra_floors, extra_bedrooms, extra_living = calculate_room_multipliers(bedrooms, washrooms, living, floors)

    # Apply soil factor
    soil_factor = SOIL_FACTOR[soil_type]

    # --------------------
    # Grey quantities
    # --------------------

    steel_kg = total_covered * (BASE_NORMS["steel_kg"] + room_multipliers["steel_kg"]) * soil_factor + addon_impacts["steel_kg"]
    cement_bags = total_covered * (BASE_NORMS["cement_bags"] + room_multipliers["cement_bags"]) * soil_factor + addon_impacts["cement_bags"]
    sand_cft = total_covered * (BASE_NORMS["sand_cft"] + room_multipliers["sand_cft"]) * soil_factor + addon_impacts["sand_cft"]
    crush_cft = total_covered * (BASE_NORMS["crush_cft"] + room_multipliers["crush_cft"]) * soil_factor + addon_impacts["crush_cft"]
    bricks = total_covered * (BASE_NORMS["bricks"] + room_multipliers["bricks"])

    # --------------------
    # Finish quantities
    # --------------------

    wiring_points = total_covered * (BASE_NORMS["wiring_points"] + room_multipliers["wiring_points"])
    plumbing_points = total_covered * (BASE_NORMS["plumbing_points"] + room_multipliers["plumbing_points"])
    tiles_area = total_covered * (BASE_NORMS["tiles_sqft"] + room_multipliers["tiles_sqft"]) + addon_impacts["tiles_sqft"]
    paint_area = total_covered * (BASE_NORMS["paint_sqft"] + room_multipliers["paint_sqft"]) + addon_impacts["paint_sqft"]
    ceiling_area = total_covered * BASE_NORMS["ceiling_sqft"]
    wood_area = total_covered * (BASE_NORMS["wood_sqft"] + room_multipliers["wood_sqft"])

    # --------------------
    # Grey material cost
    # --------------------

    steel_cost = steel_kg * MATERIAL_PRICES["steel"][brands["steel"]]
    cement_cost = cement_bags * MATERIAL_PRICES["cement"][brands["cement"]]
    sand_cost = sand_cft * MATERIAL_PRICES["sand"][brands["sand"]]
    crush_cost = crush_cft * MATERIAL_PRICES["crush"][brands["crush"]]
    brick_cost = bricks * MATERIAL_PRICES["bricks"][brands["bricks"]]

    grey_material_cost = steel_cost + cement_cost + sand_cost + crush_cost + brick_cost

    # --------------------
    # Finish material cost
    # --------------------

    wire_cost = wiring_points * MATERIAL_PRICES["wire"][brands["wire"]]
    tile_cost = tiles_area * MATERIAL_PRICES["tiles"][brands["tiles"]]
    paint_cost = paint_area * MATERIAL_PRICES["paint"][brands["paint"]]
    wood_cost = wood_area * MATERIAL_PRICES["wood"][brands["wood"]]

    sanitary_cost = washrooms * MATERIAL_PRICES["sanitary"][brands["sanitary"]]

    # Misc finishing costs
    plumbing_cost = plumbing_points * 1800
    electrical_misc = total_covered * 70
    waterproofing_multiplier = 1 + addon_impacts["waterproofing_multiplier"] + (washrooms * ROOM_MULTIPLIERS["per_washroom"]["waterproofing_multiplier"])
    waterproofing_cost = total_covered * 80 * waterproofing_multiplier
    metal_glass_misc = total_covered * 120 + addon_impacts["metal_glass_cost"]
    paint_labour_material = total_covered * 150

    finish_material_cost = (
        wire_cost
        + tile_cost
        + paint_cost
        + wood_cost
        + sanitary_cost
        + plumbing_cost
        + electrical_misc
        + waterproofing_cost
        + metal_glass_misc
        + paint_labour_material
    )

    # Apply quality multiplier to finishing
    finish_material_cost *= QUALITY_MULTIPLIER[quality_level]

    # --------------------
    # City factor (applied to both)
    # --------------------

    city_factor = CITY_FACTOR.get(city, 1.00)

    grey_material_cost *= city_factor
    finish_material_cost *= city_factor

    # --------------------
    # LABOUR COSTS (Direct wages to workers)
    # --------------------

    labour_grey = total_covered * LABOUR_RATE_GREY * (1 + extra_floors * ROOM_MULTIPLIERS["per_extra_floor"]["labour_multiplier"])
    labour_finish = total_covered * LABOUR_RATE_FINISH * (1 + extra_bedrooms * 0.03 + washrooms * 0.02 + extra_living * 0.02)

    # Apply city factor to labour
    labour_grey *= city_factor
    labour_finish *= city_factor

    # --------------------
    # CONTRACTOR/DEVELOPMENT FEES
    # --------------------

    grey_contractor_fee = total_covered * GREY_CONTRACTOR_FEE * city_factor
    finish_contractor_fee = total_covered * FINISH_CONTRACTOR_FEE * city_factor

    # --------------------
    # Add-ons
    # --------------------

    addon_cost = 0
    solar_addon_cost = 0
    solar_details = None
    insulation_addon_cost = 0
    ceiling_addon_cost = 0

    if solar:
        solar_system = SOLAR_SYSTEMS.get(str(plot), SOLAR_SYSTEMS["5"])
        solar_addon_cost = solar_system["cost"]
        solar_details = {
            "cost": solar_system["cost"],
            "kw_range": solar_system["kw_range"],
            "kw_min": solar_system["kw_min"],
            "kw_max": solar_system["kw_max"],
            "panels": solar_system["panels"],
            "panel_watt": solar_system["panel_watt"],
            "battery": solar_system["battery"],
            "ac_coverage": solar_system["ac_coverage"]
        }
        addon_cost += solar_addon_cost

    if insulation and insulation_type:
        insulation_addon_cost = total_covered * INSULATION_RATE[insulation_type] * city_factor
        addon_cost += insulation_addon_cost

    if ceiling_type:
        ceiling_addon_cost = total_covered * MATERIAL_PRICES["ceiling"][ceiling_type] * city_factor
        addon_cost += ceiling_addon_cost

    # --------------------
    # TOTALS
    # --------------------

    grey_total = grey_material_cost + labour_grey + grey_contractor_fee
    finish_total = finish_material_cost + labour_finish + finish_contractor_fee + addon_cost
    grand_total = grey_total + finish_total
    cost_per_sqft = grand_total / total_covered

    # Quantity outputs
    quantities_output = {
        "steel_kg": round(steel_kg, 1),
        "cement_bags": round(cement_bags, 1),
        "sand_cft": round(sand_cft, 1),
        "crush_cft": round(crush_cft, 1),
        "bricks": int(bricks),
        "wiring_points": int(wiring_points),
        "plumbing_points": int(plumbing_points),
        "tiles_sqft": round(tiles_area, 1),
        "paint_sqft": round(paint_area, 1),
        "wood_sqft": round(wood_area, 1)
    }

    if solar and solar_details:
        quantities_output["solar_system"] = solar_details

    if insulation and insulation_type:
        quantities_output["insulation_area_sqft"] = round(total_covered, 1)
        quantities_output["insulation_type"] = insulation_type

    return {
        "total_covered_area_sqft": round(total_covered, 2),
        "plot_size_marla": plot,
        "quantities": quantities_output,
        "grey_total": round(grey_total, 0),
        "finish_total": round(finish_total, 0),
        "grand_total": round(grand_total, 0),
        "difference": round(finish_total - grey_total, 0),
        "cost_per_sqft": round(cost_per_sqft, 0),
        "solar_details": solar_details,
        "costs": {
            "grey_material": round(grey_material_cost, 0),
            "grey_labour": round(labour_grey, 0),
            "grey_contractor": round(grey_contractor_fee, 0),
            "finish_material": round(finish_material_cost, 0),
            "finish_labour": round(labour_finish, 0),
            "finish_contractor": round(finish_contractor_fee, 0),
            "solar_addon": round(solar_addon_cost, 0),
            "insulation_addon": round(insulation_addon_cost, 0),
            "ceiling_addon": round(ceiling_addon_cost, 0),
        }
    }
```

```python
def ask_bool(msg):
    return input(msg + " (y/n): ").strip().lower() == "y"

def get_int(msg, min_val, max_val):
    while True:
        try:
            val = int(input(msg))
            if val < min_val or val > max_val:
                print(f"⚠️ Enter between {min_val} and {max_val}")
            else:
                return val
        except:
            print("⚠️ Invalid number")

print("\n" + "="*60)
print("🏗️  PAKISTAN CONSTRUCTION COST ENGINE")
print("="*60)
print("\n📋 RULES:")
print("   • Max floors: 3")
print("   • 5 marla: No basement")
print("   • Bathrooms: min = half bedrooms, max = equal bedrooms")
print("   • Prices reflect well-developed housing societies (DHA, Bahria, etc.)\n")

print("📐 PLOT CAPACITIES:")
print(f"   • 5 marla:  Max {PLOT_RULES['5']['max_bed_total']} beds ({PLOT_RULES['5']['max_bed_per_floor']}/floor), {PLOT_RULES['5']['max_living_total']} living")
print(f"   • 10 marla: Max {PLOT_RULES['10']['max_bed_total']} beds ({PLOT_RULES['10']['max_bed_per_floor']}/floor), {PLOT_RULES['10']['max_living_total']} living")
print(f"   • 15 marla: Max {PLOT_RULES['15']['max_bed_total']} beds ({PLOT_RULES['15']['max_bed_per_floor']}/floor), {PLOT_RULES['15']['max_living_total']} living")
print(f"   • 20 marla: Max {PLOT_RULES['20']['max_bed_total']} beds ({PLOT_RULES['20']['max_bed_per_floor']}/floor), {PLOT_RULES['20']['max_living_total']} living\n")

# -------- Plot --------
print("Available plots:", list(PLOT_RULES.keys()))
plot = input("Plot size (5/10/15/20): ").strip()

floors = get_int("Number of floors (1-3): ", 1, 3)

# Get limits for this plot
rule = PLOT_RULES[plot]
max_beds_total = rule["max_bed_total"]
max_beds_per_floor = rule["max_bed_per_floor"]
max_living_total = rule["max_living_total"]
max_living_per_floor = rule["max_living_per_floor"]

print(f"\n✅ For {plot} marla: Max {max_beds_total} bedrooms total, {max_beds_per_floor} per floor")

# -------- Room input per floor --------
beds = 0
living = 0

for f in range(1, floors+1):
    remaining_beds = max_beds_total - beds
    remaining_living = max_living_total - living

    if remaining_beds <= 0:
        print(f"\n⚠️ Cannot add Floor {f} - already reached max bedrooms ({max_beds_total})")
        break

    max_beds_this_floor = min(max_beds_per_floor, remaining_beds)
    max_living_this_floor = min(max_living_per_floor, remaining_living)

    print(f"\n🏠 Floor {f}")
    print(f"   Remaining capacity: {remaining_beds} beds, {remaining_living} living")
    print(f"   Max this floor: {max_beds_this_floor} beds, {max_living_this_floor} living")

    b = get_int(f"   Bedrooms on floor {f}: ", 1, max_beds_this_floor)
    l = get_int(f"   Living rooms on floor {f}: ", 1, max_living_this_floor)

    beds += b
    living += l

print(f"\n✅ TOTAL: {beds} bedrooms, {living} living rooms across {floors} floor(s)")

# -------- Bathrooms --------
min_bath = max(1, beds//2)
max_bath = beds

print(f"\n🚿 Bathrooms allowed: {min_bath} to {max_bath} (based on {beds} bedrooms)")
washrooms = get_int("Number of washrooms: ", min_bath, max_bath)

# -------- City --------
print("\n🏙️ CITIES (with cost impact):")
for city, factor in CITY_FACTOR.items():
    impact = "Base" if factor == 1.00 else f"+{(factor-1)*100:.0f}%" if factor > 1 else f"{(factor-1)*100:.0f}%"
    print(f"   • {city.title()}: {impact}")
city = input("\nSelect city: ").strip().lower()

if city not in CITY_FACTOR:
    print(f"⚠️ City '{city}' not found. Using Lahore as default.")
    city = "lahore"

# -------- Soil --------
print("\n🌱 SOIL TYPES:")
print("   • Normal: Standard foundation")
print("   • Soft: +8% material (needs deeper foundation)")
print("   • Filled: +15% material (needs extra strengthening)")
soil = input("Soil type (normal/soft/filled): ").strip().lower()

# -------- Quality --------
print("\n✨ QUALITY LEVELS:")
print("   • Standard: Regular finishes (100%)")
print("   • Premium: Better materials (+18%)")
print("   • Luxury: High-end finishes (+42%)")
quality = input("Quality level (standard/premium/luxury): ").strip().lower()

# -------- Brands --------
print("\n--- SELECT BRANDS ---")
brands = {}

brand_categories = ["steel", "cement", "sand", "crush", "bricks", "wire", "tiles", "paint", "wood", "sanitary"]
for k in brand_categories:
    print(f"\n{k.upper()}:")
    print(f"   Options: {', '.join(MATERIAL_PRICES[k].keys())}")
    brands[k] = input(f"   Choose {k}: ").strip().lower()

# -------- Extras --------
print("\n--- ADDITIONAL FEATURES ---")
porch = ask_bool("Include porch? (+5% area)")
balcony = ask_bool("Include balcony? (+3% area)")

print("\n🏠 CEILING TYPES:")
print(f"   Options: {', '.join(MATERIAL_PRICES['ceiling'].keys())}")
ceiling = input("Ceiling type (optional, press Enter to skip): ").strip() or None

# -------- Solar System with detailed info --------
print("\n☀️ SOLAR SYSTEM OPTIONS:")
for marla, details in SOLAR_SYSTEMS.items():
    print(f"\n   {marla} MARLA SYSTEM:")
    print(f"      • Capacity: {details['kw_range']}")
    print(f"      • Panels: {details['panels']} x {details['panel_watt']}W")
    print(f"      • Battery: {details['battery'].upper()}")
    print(f"      • Coverage: {details['ac_coverage']}")
    print(f"      • Cost: PKR {details['cost']:,}")

solar = ask_bool("\nAdd Solar System?")

# -------- Insulation --------
insulation = ask_bool("\n🛡️ Add Thermal Insulation?")
ins_type = None

if insulation:
    print(f"\n   Insulation types: {', '.join(INSULATION_RATE.keys())}")
    print("   • Basic: PKR 120/sqft")
    print("   • Thermal: PKR 180/sqft")
    print("   • Premium: PKR 250/sqft")
    ins_type = input("   Choose insulation type: ").strip().lower()

# -------- Basement (Only for 10+ marla) --------
if plot == "5":
    print("\n⚠️ Basement NOT allowed for 5 marla plots")
    basement = False
else:
    basement = ask_bool("\n🏗️ Include Basement?")

# -------- Calculation mode --------
print("\n" + "="*60)
print("📊 CALCULATION MODE:")
print("   1. Grey Structure Only (walls, roof, foundation)")
print("   2. Full House Only (complete finishing)")
print("   3. Both + Detailed Breakdown")
choice = input("Enter 1, 2, or 3: ").strip()

# -------- Run calculation --------
result = estimate_house(
    plot=plot,
    floors=floors,
    bedrooms=beds,
    washrooms=washrooms,
    living=living,
    city=city,
    soil_type=soil,
    quality_level=quality,
    brands=brands,
    porch=porch,
    balcony=balcony,
    solar=solar,
    insulation=insulation,
    insulation_type=ins_type,
    ceiling_type=ceiling
)

# -------- Display results --------
print("\n" + "="*60)
print("📋 CONSTRUCTION COST ESTIMATE")
print("="*60)
print(f"\n📍 Location: {city.title()} (City factor: {CITY_FACTOR[city]})")
print(f"📐 Total covered area: {result['total_covered_area_sqft']:,.0f} sqft")
print(f"🏠 Configuration: {beds} bedrooms, {washrooms} washrooms, {living} living rooms")
print(f"📏 Plot size: {plot} marla")

# Display Solar details if added
if solar and result['solar_details']:
    print("\n☀️ SOLAR SYSTEM DETAILS:")
    solar_d = result['solar_details']
    print(f"   • System Capacity: {solar_d['kw_range']}")
    print(f"   • Solar Panels: {solar_d['panels']} x {solar_d['panel_watt']}W = {solar_d['panels'] * solar_d['panel_watt'] / 1000:.1f} KW total")
    print(f"   • Battery Backup: {solar_d['battery'].upper()}")
    print(f"   • Coverage: {solar_d['ac_coverage']}")
    print(f"   • Cost: PKR {solar_d['cost']:,}")

if choice == "1":
    print("\n🏗️ GREY STRUCTURE COST")
    print(f"   Total: PKR {result['grey_total']:,}")
    print(f"   Rate: PKR {result['grey_total']/result['total_covered_area_sqft']:,.0f}/sqft")
    print("\n   Breakdown:")
    print(f"   • Materials (you buy): PKR {result['costs']['grey_material']:,}")
    print(f"   • Labour (worker wages): PKR {result['costs']['grey_labour']:,}")
    print(f"   • Contractor (supervision + profit): PKR {result['costs']['grey_contractor']:,}")

elif choice == "2":
    print("\n🏠 FULL HOUSE COST")
    print(f"   Total: PKR {result['finish_total']:,}")
    print(f"   Rate: PKR {result['finish_total']/result['total_covered_area_sqft']:,.0f}/sqft")
    print("\n   Breakdown:")
    print(f"   • Materials (you buy): PKR {result['costs']['finish_material']:,}")
    print(f"   • Labour (worker wages): PKR {result['costs']['finish_labour']:,}")
    print(f"   • Contractor (supervision + profit): PKR {result['costs']['finish_contractor']:,}")
    if solar or insulation or ceiling:
        print(f"\n   Add-ons Included:")
        if solar:
            print(f"   • Solar System: PKR {result['costs']['solar_addon']:,}")
        if insulation:
            print(f"   • Insulation: PKR {result['costs']['insulation_addon']:,}")
        if ceiling:
            print(f"   • Ceiling: PKR {result['costs']['ceiling_addon']:,}")

else:  # choice 3 - Detailed
    print("\n📦 MATERIAL QUANTITIES & INDIVIDUAL COSTS:")
    print("   " + "-"*40)

    # Calculate individual material costs for display
    steel_kg = result['quantities'].get('steel_kg', 0)
    cement_bags = result['quantities'].get('cement_bags', 0)
    sand_cft = result['quantities'].get('sand_cft', 0)
    crush_cft = result['quantities'].get('crush_cft', 0)
    bricks = result['quantities'].get('bricks', 0)

    # Get material prices from brands
    steel_price = MATERIAL_PRICES["steel"][brands["steel"]]
    cement_price = MATERIAL_PRICES["cement"][brands["cement"]]
    sand_price = MATERIAL_PRICES["sand"][brands["sand"]]
    crush_price = MATERIAL_PRICES["crush"][brands["crush"]]
    brick_price = MATERIAL_PRICES["bricks"][brands["bricks"]]

    # Calculate individual grey material costs
    steel_cost_individual = steel_kg * steel_price
    cement_cost_individual = cement_bags * cement_price
    sand_cost_individual = sand_cft * sand_price
    crush_cost_individual = crush_cft * crush_price
    brick_cost_individual = bricks * brick_price

    # Calculate finishing individual costs
    wiring_points = result['quantities'].get('wiring_points', 0)
    plumbing_points = result['quantities'].get('plumbing_points', 0)
    tiles_sqft = result['quantities'].get('tiles_sqft', 0)
    paint_sqft = result['quantities'].get('paint_sqft', 0)
    wood_sqft = result['quantities'].get('wood_sqft', 0)

    wire_price = MATERIAL_PRICES["wire"][brands["wire"]]
    tile_price = MATERIAL_PRICES["tiles"][brands["tiles"]]
    paint_price = MATERIAL_PRICES["paint"][brands["paint"]]
    wood_price = MATERIAL_PRICES["wood"][brands["wood"]]
    sanitary_price = MATERIAL_PRICES["sanitary"][brands["sanitary"]]

    wire_cost_individual = wiring_points * wire_price
    tile_cost_individual = tiles_sqft * tile_price
    paint_cost_individual = paint_sqft * paint_price
    wood_cost_individual = wood_sqft * wood_price
    sanitary_cost_individual = sanitary_price * washrooms

    # Display quantities first
    print("\n   📊 MATERIAL QUANTITIES:")
    for k, v in result["quantities"].items():
        if k == "solar_system":
            continue
        if isinstance(v, (int, float)):
            if isinstance(v, float):
                print(f"      • {k.replace('_', ' ').title()}: {v:,.1f}")
            else:
                print(f"      • {k.replace('_', ' ').title()}: {v:,}")
        else:
            print(f"      • {k.replace('_', ' ').title()}: {v}")

    # Display individual grey material costs
    print("\n   🏗️ GREY MATERIALS BREAKDOWN:")
    print(f"      • Steel ({steel_kg:,.1f} kg @ PKR {steel_price:,}/kg):      PKR {steel_cost_individual:,.0f}")
    print(f"      • Cement ({cement_bags:,.1f} bags @ PKR {cement_price:,}/bag):   PKR {cement_cost_individual:,.0f}")
    print(f"      • Sand ({sand_cft:,.1f} cft @ PKR {sand_price:,}/cft):        PKR {sand_cost_individual:,.0f}")
    print(f"      • Crush ({crush_cft:,.1f} cft @ PKR {crush_price:,}/cft):       PKR {crush_cost_individual:,.0f}")
    print(f"      • Bricks ({bricks:,} @ PKR {brick_price:,}/each):       PKR {brick_cost_individual:,.0f}")
    print(f"      " + "-"*35)
    print(f"      • TOTAL GREY MATERIALS:                     PKR {result['costs']['grey_material']:,}")

    # Display individual finishing material costs
    print("\n   🏠 FINISHING MATERIALS BREAKDOWN:")
    print(f"      • Wiring ({wiring_points} points @ PKR {wire_price:,}/point):   PKR {wire_cost_individual:,.0f}")
    print(f"      • Tiles ({tiles_sqft:,.1f} sqft @ PKR {tile_price:,}/sqft):     PKR {tile_cost_individual:,.0f}")
    print(f"      • Paint ({paint_sqft:,.1f} sqft @ PKR {paint_price:,}/sqft):     PKR {paint_cost_individual:,.0f}")
    print(f"      • Wood ({wood_sqft:,.1f} sqft @ PKR {wood_price:,}/sqft):       PKR {wood_cost_individual:,.0f}")
    print(f"      • Sanitary ({washrooms} sets @ PKR {sanitary_price:,}/set):   PKR {sanitary_cost_individual:,.0f}")
    print(f"      " + "-"*35)

    # Show total finishing materials
    finish_materials_total = wire_cost_individual + tile_cost_individual + paint_cost_individual + wood_cost_individual + sanitary_cost_individual
    print(f"      • TOTAL FINISH MATERIALS (basic):          PKR {finish_materials_total:,.0f}")
    print(f"      • PLUS misc items (plumbing, electrical, waterproofing, etc.)")
    print(f"      " + "-"*35)
    print(f"      • TOTAL FINISH MATERIALS (all):            PKR {result['costs']['finish_material']:,}")

    print("\n💰 COMPLETE COST BREAKDOWN:")
    print("   " + "-"*40)
    print("   GREY STRUCTURE:")
    print(f"      • Grey Materials (you buy):    PKR {result['costs']['grey_material']:,}")
    print(f"      • Grey Labour (worker wages):  PKR {result['costs']['grey_labour']:,}")
    print(f"      • Grey Contractor (fee):       PKR {result['costs']['grey_contractor']:,}")
    print("\n   FINISHING:")
    print(f"      • Finish Materials (you buy):  PKR {result['costs']['finish_material']:,}")
    print(f"      • Finish Labour (worker wages):PKR {result['costs']['finish_labour']:,}")
    print(f"      • Finish Contractor (fee):     PKR {result['costs']['finish_contractor']:,}")

    if solar or insulation or ceiling:
        print("\n   ADD-ONS:")
        if solar:
            print(f"      • Solar System:               PKR {result['costs']['solar_addon']:,}")
        if insulation:
            print(f"      • Insulation:                 PKR {result['costs']['insulation_addon']:,}")
        if ceiling:
            print(f"      • Ceiling:                    PKR {result['costs']['ceiling_addon']:,}")

    print("\n" + "-"*40)
    print(f"🏗️ GREY STRUCTURE TOTAL:     PKR {result['grey_total']:,}")
    print(f"🏠 FINISHING TOTAL:          PKR {result['finish_total']:,}")
    print(f"📊 DIFFERENCE:               PKR {result['difference']:,}")
    print(f"🎯 GRAND TOTAL:              PKR {result['grand_total']:,}")
    print(f"💰 COST PER SQFT:            PKR {result['cost_per_sqft']:,}")

print("\n" + "="*60)
print("✅ Estimate complete!")
print("="*60)
```

```python
import math

# =========================================================
# Pakistan Parametric Construction Cost Engine (v3)
# PRODUCTION VERSION - Reads ALL rates from Config (Cell 0)
# =========================================================

# IMPORTANT: Cell 0 (Config Manager) must be run before this cell!

ENGINE_METADATA = {
    "engine": "Pakistan Parametric Cost Engine",
    "version": "3.0",
    "currency": "PKR",
    "area_unit": "sqft",
    "target_market": "Well-developed housing societies (DHA, Bahria, Gulberg, etc.)"
}


# -----------------------
# ALL CONFIGURATION FROM CELL 0 (No hardcoded values)
# -----------------------

# Get configuration from Cell 0
config = Config.get_config()

# Plot rules
PLOT_RULES = config["plot_rules"]

# City factors
CITY_FACTOR = config["city_factors"]

# Soil factors
SOIL_FACTOR = config["soil_factors"]

# Quality multipliers
QUALITY_MULTIPLIER = config["quality_multipliers"]

# Labour & Contractor Rates
LABOUR_RATE_GREY = config["labour_rates"]["grey_structure"]
LABOUR_RATE_FINISH = config["labour_rates"]["finishing"]
GREY_CONTRACTOR_FEE = config["contractor_fees"]["grey_structure"]
FINISH_CONTRACTOR_FEE = config["contractor_fees"]["finishing"]

# Base consumption norms
BASE_NORMS = {
    "steel_kg": config["base_consumption"]["grey_structure"]["steel_kg"],
    "cement_bags": config["base_consumption"]["grey_structure"]["cement_bags"],
    "sand_cft": config["base_consumption"]["grey_structure"]["sand_cft"],
    "crush_cft": config["base_consumption"]["grey_structure"]["crush_cft"],
    "bricks": config["base_consumption"]["grey_structure"]["bricks"],
    "wiring_points": config["base_consumption"]["finishing"]["wiring_points"],
    "plumbing_points": config["base_consumption"]["finishing"]["plumbing_points"],
    "tiles_sqft": config["base_consumption"]["finishing"]["tiles_sqft"],
    "paint_sqft": config["base_consumption"]["finishing"]["paint_sqft"],
    "ceiling_sqft": config["base_consumption"]["finishing"]["ceiling_sqft"],
    "wood_sqft": config["base_consumption"]["finishing"]["wood_sqft"]
}

# Room multipliers
ROOM_MULTIPLIERS = config["room_multipliers"]

# Add-on multipliers
ADDON_MULTIPLIERS = config["addon_multipliers"]

# Material prices
MATERIAL_PRICES = config["material_prices"]

# Insulation rates
INSULATION_RATE = config["insulation_rates"]


# =========================================================
# Helper Functions
# =========================================================

def calculate_room_multipliers(bedrooms, washrooms, living, floors):
    """Calculate material additions based on room configuration"""

    extra_bedrooms = max(0, bedrooms - 2)
    extra_living = max(0, living - 1)
    extra_floors = max(0, floors - 1)

    multipliers = {
        "steel_kg": 0, "cement_bags": 0, "sand_cft": 0, "crush_cft": 0,
        "bricks": 0, "wiring_points": 0, "plumbing_points": 0,
        "tiles_sqft": 0, "paint_sqft": 0, "wood_sqft": 0
    }

    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_extra_bedroom"]:
            multipliers[key] += extra_bedrooms * ROOM_MULTIPLIERS["per_extra_bedroom"][key]

    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_washroom"]:
            multipliers[key] += washrooms * ROOM_MULTIPLIERS["per_washroom"][key]

    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_extra_living"]:
            multipliers[key] += extra_living * ROOM_MULTIPLIERS["per_extra_living"][key]

    for key in multipliers:
        if key in ROOM_MULTIPLIERS["per_extra_floor"]:
            multipliers[key] += extra_floors * ROOM_MULTIPLIERS["per_extra_floor"][key]

    return multipliers, extra_floors, extra_bedrooms, extra_living


def calculate_addon_impacts(has_porch, has_balcony, base_area):
    """Calculate material additions from porch and balcony"""

    porch_area = 0
    balcony_area = 0
    addon_multipliers = {
        "steel_kg": 0, "cement_bags": 0, "sand_cft": 0, "crush_cft": 0,
        "tiles_sqft": 0, "paint_sqft": 0, "metal_glass_cost": 0,
        "waterproofing_multiplier": 0
    }

    if has_porch:
        porch_area = base_area * ADDON_MULTIPLIERS["porch"]["area_multiplier"]
        addon_multipliers["steel_kg"] += porch_area * ADDON_MULTIPLIERS["porch"]["steel_kg"]
        addon_multipliers["cement_bags"] += porch_area * ADDON_MULTIPLIERS["porch"]["cement_bags"]
        addon_multipliers["sand_cft"] += porch_area * ADDON_MULTIPLIERS["porch"]["sand_cft"]
        addon_multipliers["crush_cft"] += porch_area * ADDON_MULTIPLIERS["porch"]["crush_cft"]
        addon_multipliers["tiles_sqft"] += porch_area * ADDON_MULTIPLIERS["porch"]["tiles_sqft"]
        addon_multipliers["paint_sqft"] += porch_area * ADDON_MULTIPLIERS["porch"]["paint_sqft"]
        addon_multipliers["metal_glass_cost"] += porch_area * ADDON_MULTIPLIERS["porch"]["metal_glass"]
        addon_multipliers["waterproofing_multiplier"] += ADDON_MULTIPLIERS["porch"]["waterproofing"]

    if has_balcony:
        balcony_area = base_area * ADDON_MULTIPLIERS["balcony"]["area_multiplier"]
        addon_multipliers["steel_kg"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["steel_kg"]
        addon_multipliers["cement_bags"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["cement_bags"]
        addon_multipliers["sand_cft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["sand_cft"]
        addon_multipliers["crush_cft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["crush_cft"]
        addon_multipliers["tiles_sqft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["tiles_sqft"]
        addon_multipliers["paint_sqft"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["paint_sqft"]
        addon_multipliers["metal_glass_cost"] += balcony_area * ADDON_MULTIPLIERS["balcony"]["metal_glass"]
        addon_multipliers["waterproofing_multiplier"] += ADDON_MULTIPLIERS["balcony"]["waterproofing"]

    total_addon_area = porch_area + balcony_area

    return addon_multipliers, total_addon_area


# =========================================================
# Core estimator (NO SOLAR)
# =========================================================

def estimate_house(
    plot,
    floors,
    bedrooms,
    washrooms,
    living,
    city,
    soil_type,
    quality_level,
    brands,
    porch=False,
    balcony=False,
    insulation=False,
    insulation_type=None,
    ceiling_type=None
):

    if plot not in PLOT_RULES:
        raise ValueError("Invalid plot size")

    rule = PLOT_RULES[plot]

    if bedrooms > rule["max_bed_total"]:
        raise ValueError(f"Bedrooms ({bedrooms}) exceed plot limit ({rule['max_bed_total']})")
    if washrooms > rule["max_wash"]:
        raise ValueError(f"Washrooms ({washrooms}) exceed plot limit ({rule['max_wash']})")
    if living > rule["max_living_total"]:
        raise ValueError(f"Living rooms ({living}) exceed plot limit ({rule['max_living_total']})")

    plot_sqft = rule["plot_sqft"]
    coverage = rule["coverage"]
    covered_per_floor = plot_sqft * coverage
    base_area = covered_per_floor * floors

    addon_impacts, addon_area = calculate_addon_impacts(porch, balcony, base_area)
    total_covered = base_area + addon_area

    room_multipliers, extra_floors, extra_bedrooms, extra_living = calculate_room_multipliers(bedrooms, washrooms, living, floors)

    soil_factor = SOIL_FACTOR[soil_type]

    # Grey quantities
    steel_kg = total_covered * (BASE_NORMS["steel_kg"] + room_multipliers["steel_kg"]) * soil_factor + addon_impacts["steel_kg"]
    cement_bags = total_covered * (BASE_NORMS["cement_bags"] + room_multipliers["cement_bags"]) * soil_factor + addon_impacts["cement_bags"]
    sand_cft = total_covered * (BASE_NORMS["sand_cft"] + room_multipliers["sand_cft"]) * soil_factor + addon_impacts["sand_cft"]
    crush_cft = total_covered * (BASE_NORMS["crush_cft"] + room_multipliers["crush_cft"]) * soil_factor + addon_impacts["crush_cft"]
    bricks = total_covered * (BASE_NORMS["bricks"] + room_multipliers["bricks"])

    # Finish quantities
    wiring_points = total_covered * (BASE_NORMS["wiring_points"] + room_multipliers["wiring_points"])
    plumbing_points = total_covered * (BASE_NORMS["plumbing_points"] + room_multipliers["plumbing_points"])
    tiles_area = total_covered * (BASE_NORMS["tiles_sqft"] + room_multipliers["tiles_sqft"]) + addon_impacts["tiles_sqft"]
    paint_area = total_covered * (BASE_NORMS["paint_sqft"] + room_multipliers["paint_sqft"]) + addon_impacts["paint_sqft"]
    ceiling_area = total_covered * BASE_NORMS["ceiling_sqft"]
    wood_area = total_covered * (BASE_NORMS["wood_sqft"] + room_multipliers["wood_sqft"])

    # Grey material cost
    steel_cost = steel_kg * MATERIAL_PRICES["steel"][brands["steel"]]
    cement_cost = cement_bags * MATERIAL_PRICES["cement"][brands["cement"]]
    sand_cost = sand_cft * MATERIAL_PRICES["sand"][brands["sand"]]
    crush_cost = crush_cft * MATERIAL_PRICES["crush"][brands["crush"]]
    brick_cost = bricks * MATERIAL_PRICES["bricks"][brands["bricks"]]

    grey_material_cost = steel_cost + cement_cost + sand_cost + crush_cost + brick_cost

    # Finish material cost - CORRECTED
    wire_cost = wiring_points * MATERIAL_PRICES["wire"][brands["wire"]]
    tile_cost = tiles_area * MATERIAL_PRICES["tiles"][brands["tiles"]]

    # CORRECTED PAINT COST (per gallon, 1 gallon = 250 sqft for 2 coats)
    paint_price_per_gallon = MATERIAL_PRICES["paint"][brands["paint"]]
    gallons_needed = paint_area / 250
    paint_cost = gallons_needed * paint_price_per_gallon

    # CORRECTED WOOD COST (per cubic foot to per sqft installed)
    wood_price_per_cubic_ft = MATERIAL_PRICES["wood"][brands["wood"]]
    wood_price_per_sqft_installed = wood_price_per_cubic_ft / 3.5
    wood_cost = wood_area * wood_price_per_sqft_installed

    sanitary_cost = washrooms * MATERIAL_PRICES["sanitary"][brands["sanitary"]]

    # Misc finishing costs (REDUCED to realistic rates)
    plumbing_cost = plumbing_points * 1500
    electrical_misc = total_covered * 35
    waterproofing_multiplier = 1 + addon_impacts["waterproofing_multiplier"] + (washrooms * ROOM_MULTIPLIERS["per_washroom"]["waterproofing_multiplier"])
    waterproofing_cost = total_covered * 60 * waterproofing_multiplier
    metal_glass_misc = total_covered * 100 + addon_impacts["metal_glass_cost"]
    paint_labour_material = total_covered * 100

    finish_material_cost = (
        wire_cost + tile_cost + paint_cost + wood_cost + sanitary_cost +
        plumbing_cost + electrical_misc + waterproofing_cost + metal_glass_misc + paint_labour_material
    )

    finish_material_cost *= QUALITY_MULTIPLIER[quality_level]

    # City factor
    city_factor = CITY_FACTOR.get(city, 1.00)
    grey_material_cost *= city_factor
    finish_material_cost *= city_factor

    # Labour costs
    labour_grey = total_covered * LABOUR_RATE_GREY * (1 + extra_floors * ROOM_MULTIPLIERS["per_extra_floor"]["labour_multiplier"])
    labour_finish = total_covered * LABOUR_RATE_FINISH * (1 + extra_bedrooms * 0.03 + washrooms * 0.02 + extra_living * 0.02)

    labour_grey *= city_factor
    labour_finish *= city_factor

    # Contractor fees
    grey_contractor_fee = total_covered * GREY_CONTRACTOR_FEE * city_factor
    finish_contractor_fee = total_covered * FINISH_CONTRACTOR_FEE * city_factor

    # Add-ons (Insulation only, NO SOLAR)
    addon_cost = 0
    insulation_addon_cost = 0
    ceiling_addon_cost = 0

    if insulation and insulation_type:
        insulation_addon_cost = total_covered * INSULATION_RATE[insulation_type] * city_factor
        addon_cost += insulation_addon_cost

    if ceiling_type:
        ceiling_addon_cost = total_covered * MATERIAL_PRICES["ceiling"][ceiling_type] * city_factor
        addon_cost += ceiling_addon_cost

    # Totals
    grey_total = grey_material_cost + labour_grey + grey_contractor_fee
    finish_total = finish_material_cost + labour_finish + finish_contractor_fee + addon_cost
    grand_total = grey_total + finish_total
    cost_per_sqft = grand_total / total_covered

    # Quantity outputs
    quantities_output = {
        "steel_kg": round(steel_kg, 1),
        "cement_bags": round(cement_bags, 1),
        "sand_cft": round(sand_cft, 1),
        "crush_cft": round(crush_cft, 1),
        "bricks": int(bricks),
        "wiring_points": int(wiring_points),
        "plumbing_points": int(plumbing_points),
        "tiles_sqft": round(tiles_area, 1),
        "paint_sqft": round(paint_area, 1),
        "wood_sqft": round(wood_area, 1)
    }

    if insulation and insulation_type:
        quantities_output["insulation_area_sqft"] = round(total_covered, 1)
        quantities_output["insulation_type"] = insulation_type

    return {
        "total_covered_area_sqft": round(total_covered, 2),
        "plot_size_marla": plot,
        "quantities": quantities_output,
        "grey_total": round(grey_total, 0),
        "finish_total": round(finish_total, 0),
        "grand_total": round(grand_total, 0),
        "difference": round(finish_total - grey_total, 0),
        "cost_per_sqft": round(cost_per_sqft, 0),
        "costs": {
            "grey_material": round(grey_material_cost, 0),
            "grey_labour": round(labour_grey, 0),
            "grey_contractor": round(grey_contractor_fee, 0),
            "finish_material": round(finish_material_cost, 0),
            "finish_labour": round(labour_finish, 0),
            "finish_contractor": round(finish_contractor_fee, 0),
            "insulation_addon": round(insulation_addon_cost, 0),
            "ceiling_addon": round(ceiling_addon_cost, 0),
        }
    }
```

```python
def ask_bool(msg):
    return input(msg + " (y/n): ").strip().lower() == "y"

def get_int(msg, min_val, max_val):
    while True:
        try:
            val = int(input(msg))
            if val < min_val or val > max_val:
                print(f"⚠️ Enter between {min_val} and {max_val}")
            else:
                return val
        except:
            print("⚠️ Invalid number")

print("\n" + "="*60)
print("🏗️  PAKISTAN CONSTRUCTION COST ENGINE")
print("="*60)
print("\n📋 RULES:")
print("   • Max floors: 3")
print("   • 5 marla: No basement")
print("   • Bathrooms: min = half bedrooms, max = equal bedrooms")
print("   • Prices reflect well-developed housing societies (DHA, Bahria, etc.)\n")

print("📐 PLOT CAPACITIES:")
print(f"   • 5 marla:  Max {PLOT_RULES['5']['max_bed_total']} beds ({PLOT_RULES['5']['max_bed_per_floor']}/floor), {PLOT_RULES['5']['max_living_total']} living")
print(f"   • 10 marla: Max {PLOT_RULES['10']['max_bed_total']} beds ({PLOT_RULES['10']['max_bed_per_floor']}/floor), {PLOT_RULES['10']['max_living_total']} living")
print(f"   • 15 marla: Max {PLOT_RULES['15']['max_bed_total']} beds ({PLOT_RULES['15']['max_bed_per_floor']}/floor), {PLOT_RULES['15']['max_living_total']} living")
print(f"   • 20 marla: Max {PLOT_RULES['20']['max_bed_total']} beds ({PLOT_RULES['20']['max_bed_per_floor']}/floor), {PLOT_RULES['20']['max_living_total']} living\n")

# -------- Plot --------
print("Available plots:", list(PLOT_RULES.keys()))
plot = input("Plot size (5/10/15/20): ").strip()

floors = get_int("Number of floors (1-3): ", 1, 3)

# Get limits for this plot
rule = PLOT_RULES[plot]
max_beds_total = rule["max_bed_total"]
max_beds_per_floor = rule["max_bed_per_floor"]
max_living_total = rule["max_living_total"]
max_living_per_floor = rule["max_living_per_floor"]

print(f"\n✅ For {plot} marla: Max {max_beds_total} bedrooms total, {max_beds_per_floor} per floor")

# -------- Room input per floor --------
beds = 0
living = 0

for f in range(1, floors+1):
    remaining_beds = max_beds_total - beds
    remaining_living = max_living_total - living

    if remaining_beds <= 0:
        print(f"\n⚠️ Cannot add Floor {f} - already reached max bedrooms ({max_beds_total})")
        break

    max_beds_this_floor = min(max_beds_per_floor, remaining_beds)
    max_living_this_floor = min(max_living_per_floor, remaining_living)

    print(f"\n🏠 Floor {f}")
    print(f"   Remaining capacity: {remaining_beds} beds, {remaining_living} living")
    print(f"   Max this floor: {max_beds_this_floor} beds, {max_living_this_floor} living")

    b = get_int(f"   Bedrooms on floor {f}: ", 1, max_beds_this_floor)
    l = get_int(f"   Living rooms on floor {f}: ", 1, max_living_this_floor)

    beds += b
    living += l

print(f"\n✅ TOTAL: {beds} bedrooms, {living} living rooms across {floors} floor(s)")

# -------- Bathrooms --------
min_bath = max(1, beds//2)
max_bath = beds

print(f"\n🚿 Bathrooms allowed: {min_bath} to {max_bath} (based on {beds} bedrooms)")
washrooms = get_int("Number of washrooms: ", min_bath, max_bath)

# -------- City --------
print("\n🏙️ CITIES (with cost impact):")
for city, factor in CITY_FACTOR.items():
    impact = "Base" if factor == 1.00 else f"+{(factor-1)*100:.0f}%" if factor > 1 else f"{(factor-1)*100:.0f}%"
    print(f"   • {city.title()}: {impact}")
city = input("\nSelect city: ").strip().lower()

if city not in CITY_FACTOR:
    print(f"⚠️ City '{city}' not found. Using Lahore as default.")
    city = "lahore"

# -------- Soil --------
print("\n🌱 SOIL TYPES:")
print("   • Normal: Standard foundation")
print("   • Soft: +8% material (needs deeper foundation)")
print("   • Filled: +15% material (needs extra strengthening)")
soil = input("Soil type (normal/soft/filled): ").strip().lower()

# -------- Quality --------
print("\n✨ QUALITY LEVELS:")
print("   • Standard: Regular finishes (100%)")
print("   • Premium: Better materials (+20%)")
print("   • Luxury: High-end finishes (+45%)")
quality = input("Quality level (standard/premium/luxury): ").strip().lower()

# -------- Brands --------
print("\n--- SELECT BRANDS ---")
brands = {}

brand_categories = ["steel", "cement", "sand", "crush", "bricks", "wire", "tiles", "paint", "wood", "sanitary"]
for k in brand_categories:
    print(f"\n{k.upper()}:")
    print(f"   Options: {', '.join(MATERIAL_PRICES[k].keys())}")
    brands[k] = input(f"   Choose {k}: ").strip().lower()

# -------- Extras --------
print("\n--- ADDITIONAL FEATURES ---")
porch = ask_bool("Include porch? (+5% area)")
balcony = ask_bool("Include balcony? (+3% area)")

print("\n🏠 CEILING TYPES:")
print(f"   Options: {', '.join(MATERIAL_PRICES['ceiling'].keys())}")
ceiling = input("Ceiling type (optional, press Enter to skip): ").strip() or None

# -------- Insulation --------
insulation = ask_bool("\n🛡️ Add Thermal Insulation?")
ins_type = None

if insulation:
    print(f"\n   Insulation types: {', '.join(INSULATION_RATE.keys())}")
    print("   • Basic: PKR 100/sqft")
    print("   • Thermal: PKR 150/sqft")
    print("   • Premium: PKR 220/sqft")
    ins_type = input("   Choose insulation type: ").strip().lower()

# -------- Basement (Only for 10+ marla) --------
if plot == "5":
    print("\n⚠️ Basement NOT allowed for 5 marla plots")
    basement = False
else:
    basement = ask_bool("\n🏗️ Include Basement?")

# -------- Calculation mode --------
print("\n" + "="*60)
print("📊 CALCULATION MODE:")
print("   1. Grey Structure Only (walls, roof, foundation)")
print("   2. Full House Only (complete finishing)")
print("   3. Both + Detailed Breakdown")
choice = input("Enter 1, 2, or 3: ").strip()

# -------- Run calculation (NO SOLAR) --------
result = estimate_house(
    plot=plot,
    floors=floors,
    bedrooms=beds,
    washrooms=washrooms,
    living=living,
    city=city,
    soil_type=soil,
    quality_level=quality,
    brands=brands,
    porch=porch,
    balcony=balcony,
    insulation=insulation,
    insulation_type=ins_type,
    ceiling_type=ceiling
)

# -------- Display results --------
print("\n" + "="*60)
print("📋 CONSTRUCTION COST ESTIMATE")
print("="*60)
print(f"\n📍 Location: {city.title()} (City factor: {CITY_FACTOR[city]})")
print(f"📐 Total covered area: {result['total_covered_area_sqft']:,.0f} sqft")
print(f"🏠 Configuration: {beds} bedrooms, {washrooms} washrooms, {living} living rooms")
print(f"📏 Plot size: {plot} marla")

if choice == "1":
    print("\n🏗️ GREY STRUCTURE COST")
    print(f"   Total: PKR {result['grey_total']:,}")
    print(f"   Rate: PKR {result['grey_total']/result['total_covered_area_sqft']:,.0f}/sqft")
    print("\n   Breakdown:")
    print(f"   • Materials (you buy): PKR {result['costs']['grey_material']:,}")
    print(f"   • Labour (worker wages): PKR {result['costs']['grey_labour']:,}")
    print(f"   • Contractor (supervision + profit): PKR {result['costs']['grey_contractor']:,}")

elif choice == "2":
    print("\n🏠 FULL HOUSE COST")
    print(f"   Total: PKR {result['finish_total']:,}")
    print(f"   Rate: PKR {result['finish_total']/result['total_covered_area_sqft']:,.0f}/sqft")
    print("\n   Breakdown:")
    print(f"   • Materials (you buy): PKR {result['costs']['finish_material']:,}")
    print(f"   • Labour (worker wages): PKR {result['costs']['finish_labour']:,}")
    print(f"   • Contractor (supervision + profit): PKR {result['costs']['finish_contractor']:,}")
    if insulation or ceiling:
        print(f"\n   Add-ons Included:")
        if insulation:
            print(f"   • Insulation: PKR {result['costs']['insulation_addon']:,}")
        if ceiling:
            print(f"   • Ceiling: PKR {result['costs']['ceiling_addon']:,}")

else:  # choice 3 - Detailed
    print("\n📦 MATERIAL QUANTITIES & INDIVIDUAL COSTS:")
    print("   " + "-"*40)

    steel_kg = result['quantities'].get('steel_kg', 0)
    cement_bags = result['quantities'].get('cement_bags', 0)
    sand_cft = result['quantities'].get('sand_cft', 0)
    crush_cft = result['quantities'].get('crush_cft', 0)
    bricks = result['quantities'].get('bricks', 0)

    steel_price = MATERIAL_PRICES["steel"][brands["steel"]]
    cement_price = MATERIAL_PRICES["cement"][brands["cement"]]
    sand_price = MATERIAL_PRICES["sand"][brands["sand"]]
    crush_price = MATERIAL_PRICES["crush"][brands["crush"]]
    brick_price = MATERIAL_PRICES["bricks"][brands["bricks"]]

    steel_cost_individual = steel_kg * steel_price
    cement_cost_individual = cement_bags * cement_price
    sand_cost_individual = sand_cft * sand_price
    crush_cost_individual = crush_cft * crush_price
    brick_cost_individual = bricks * brick_price

    wiring_points = result['quantities'].get('wiring_points', 0)
    plumbing_points = result['quantities'].get('plumbing_points', 0)
    tiles_sqft = result['quantities'].get('tiles_sqft', 0)
    paint_sqft = result['quantities'].get('paint_sqft', 0)
    wood_sqft = result['quantities'].get('wood_sqft', 0)

    wire_price = MATERIAL_PRICES["wire"][brands["wire"]]
    tile_price = MATERIAL_PRICES["tiles"][brands["tiles"]]
    paint_price_per_gallon = MATERIAL_PRICES["paint"][brands["paint"]]
    wood_price_per_cubic_ft = MATERIAL_PRICES["wood"][brands["wood"]]
    sanitary_price = MATERIAL_PRICES["sanitary"][brands["sanitary"]]

    wire_cost_individual = wiring_points * wire_price
    tile_cost_individual = tiles_sqft * tile_price
    gallons_needed = paint_sqft / 250
    paint_cost_individual = gallons_needed * paint_price_per_gallon
    wood_price_per_sqft_installed = wood_price_per_cubic_ft / 3.5
    wood_cost_individual = wood_sqft * wood_price_per_sqft_installed
    sanitary_cost_individual = sanitary_price * washrooms

    print("\n   📊 MATERIAL QUANTITIES:")
    for k, v in result["quantities"].items():
        if isinstance(v, (int, float)):
            if isinstance(v, float):
                print(f"      • {k.replace('_', ' ').title()}: {v:,.1f}")
            else:
                print(f"      • {k.replace('_', ' ').title()}: {v:,}")
        else:
            print(f"      • {k.replace('_', ' ').title()}: {v}")

    print("\n   🏗️ GREY MATERIALS BREAKDOWN:")
    print(f"      • Steel ({steel_kg:,.1f} kg @ PKR {steel_price:,}/kg):      PKR {steel_cost_individual:,.0f}")
    print(f"      • Cement ({cement_bags:,.1f} bags @ PKR {cement_price:,}/bag):   PKR {cement_cost_individual:,.0f}")
    print(f"      • Sand ({sand_cft:,.1f} cft @ PKR {sand_price:,}/cft):        PKR {sand_cost_individual:,.0f}")
    print(f"      • Crush ({crush_cft:,.1f} cft @ PKR {crush_price:,}/cft):       PKR {crush_cost_individual:,.0f}")
    print(f"      • Bricks ({bricks:,} @ PKR {brick_price:,}/each):       PKR {brick_cost_individual:,.0f}")
    print(f"      " + "-"*35)
    print(f"      • TOTAL GREY MATERIALS:                     PKR {result['costs']['grey_material']:,}")

    print("\n   🏠 FINISHING MATERIALS BREAKDOWN:")
    print(f"      • Wiring ({wiring_points} points @ PKR {wire_price:,}/point):   PKR {wire_cost_individual:,.0f}")
    print(f"      • Tiles ({tiles_sqft:,.1f} sqft @ PKR {tile_price:,}/sqft):     PKR {tile_cost_individual:,.0f}")
    print(f"      • Paint ({paint_sqft:,.1f} sqft @ {paint_price_per_gallon:,}/gallon): PKR {paint_cost_individual:,.0f}")
    print(f"      • Wood ({wood_sqft:,.1f} sqft @ PKR {wood_price_per_sqft_installed:,.0f}/sqft): PKR {wood_cost_individual:,.0f}")
    print(f"      • Sanitary ({washrooms} sets @ PKR {sanitary_price:,}/set):   PKR {sanitary_cost_individual:,.0f}")
    print(f"      " + "-"*35)

    finish_materials_total = wire_cost_individual + tile_cost_individual + paint_cost_individual + wood_cost_individual + sanitary_cost_individual
    print(f"      • TOTAL FINISH MATERIALS (basic):          PKR {finish_materials_total:,.0f}")
    print(f"      • PLUS misc items (plumbing, electrical, waterproofing, etc.)")
    print(f"      " + "-"*35)
    print(f"      • TOTAL FINISH MATERIALS (all):            PKR {result['costs']['finish_material']:,}")

    print("\n💰 COMPLETE COST BREAKDOWN:")
    print("   " + "-"*40)
    print("   GREY STRUCTURE:")
    print(f"      • Grey Materials (you buy):    PKR {result['costs']['grey_material']:,}")
    print(f"      • Grey Labour (worker wages):  PKR {result['costs']['grey_labour']:,}")
    print(f"      • Grey Contractor (fee):       PKR {result['costs']['grey_contractor']:,}")
    print("\n   FINISHING:")
    print(f"      • Finish Materials (you buy):  PKR {result['costs']['finish_material']:,}")
    print(f"      • Finish Labour (worker wages):PKR {result['costs']['finish_labour']:,}")
    print(f"      • Finish Contractor (fee):     PKR {result['costs']['finish_contractor']:,}")

    if insulation or ceiling:
        print("\n   ADD-ONS:")
        if insulation:
            print(f"      • Insulation:                 PKR {result['costs']['insulation_addon']:,}")
        if ceiling:
            print(f"      • Ceiling:                    PKR {result['costs']['ceiling_addon']:,}")

    print("\n" + "-"*40)
    print(f"🏗️ GREY STRUCTURE TOTAL:     PKR {result['grey_total']:,}")
    print(f"🏠 FINISHING TOTAL:          PKR {result['finish_total']:,}")
    print(f"📊 DIFFERENCE:               PKR {result['difference']:,}")
    print(f"🎯 GRAND TOTAL:              PKR {result['grand_total']:,}")
    print(f"💰 COST PER SQFT:            PKR {result['cost_per_sqft']:,}")

print("\n" + "="*60)
print("✅ Estimate complete!")
print("="*60)
```
